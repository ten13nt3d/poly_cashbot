```python
import os
from dotenv import load_dotenv
from telegram.ext import Application, CommandHandler, MessageHandler, filters, ContextTypes
from telegram import Update
import asyncio

load_dotenv()
TOKEN = os.getenv("TELEGRAM_BOT_TOKEN")

print(f"✅ Token cargado: {TOKEN[:20]}...")
```

```python
# Función /start
async def start(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user_id = update.message.from_user.id
    username = update.message.from_user.username or "Anónimo"
    print(f"\n🚀 COMANDO /start - Usuario: @{username}")
    await update.message.reply_text(
        f"¡Bienvenido al Cash Bot! 💰\n\n"
        f"Soy tu asistente de finanzas."
    )

# Función /help
async def help_command(update: Update, context: ContextTypes.DEFAULT_TYPE):
    print(f"\n❓ COMANDO /help")
    await update.message.reply_text(
        "📋 *COMANDOS DISPONIBLES*\n\n"
        "/start - Bienvenida\n"
        "/help - Esta ayuda\n"
        "/balance - Ver tu saldo actual\n"
        "/transfer - Enviar dinero\n"
        "/history - Ver historial de transacciones",
        parse_mode="Markdown"
    )

# Función /balance
async def balance(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user_id = update.message.from_user.id
    saldo = 1000.00
    print(f"\n💰 COMANDO /balance - Usuario ID: {user_id}")
    await update.message.reply_text(
        f"💰 *Tu Saldo*\n\n${saldo:.2f}",
        parse_mode="Markdown"
    )

# Función /transfer
async def transfer(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user_id = update.message.from_user.id
    print(f"\n💸 COMANDO /transfer")
    if not context.args:
        await update.message.reply_text(
            "📝 *Uso:* `/transfer @usuario cantidad`\n"
            "Ejemplo: `/transfer @juan 50`",
            parse_mode="Markdown"
        )
        return
    if len(context.args) >= 2:
        destinatario = context.args[0]
        try:
            cantidad = float(context.args[1])
            if cantidad <= 0:
                await update.message.reply_text("❌ La cantidad debe ser mayor a $0")
                return
            await update.message.reply_text(
                f"✅ *Transferencia realizada*\n\n"
                f"A: {destinatario}\n"
                f"Monto: ${cantidad:.2f}",
                parse_mode="Markdown"
            )
        except ValueError:
            await update.message.reply_text("❌ La cantidad debe ser un número válido")

# Función /history
async def history(update: Update, context: ContextTypes.DEFAULT_TYPE):
    print(f"\n📊 COMANDO /history")
    await update.message.reply_text(
        "📊 *Tu Historial*\n\n"
        "+$50 - Recibida de @juan (07/11/2025)\n"
        "-$20 - Enviada a @maria (06/11/2025)\n"
        "+$100 - Recibida de @pedro (05/11/2025)",
        parse_mode="Markdown"
    )

# Función para mensajes de texto
async def handle_message(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user_id = update.message.from_user.id
    mensaje = update.message.text
    print(f"\n📨 Mensaje de {user_id}: {mensaje}")
    await update.message.reply_text(f"Recibí: '{mensaje}'")

print("✅ Todas las funciones definidas correctamente")
```

```python
# Crear la aplicación
app = Application.builder().token(TOKEN).build()

# Registrar handlers
app.add_handler(CommandHandler("start", start))
app.add_handler(CommandHandler("help", help_command))
app.add_handler(CommandHandler("balance", balance))
app.add_handler(CommandHandler("transfer", transfer))
app.add_handler(CommandHandler("history", history))
app.add_handler(MessageHandler(filters.TEXT & ~filters.COMMAND, handle_message))

print("✅ App creada y todos los handlers registrados")
```

```python
print("🤖 INICIANDO BOT CON POLLING...")
print("="*60)
print("El bot está escuchando mensajes...")
print("Prueba: /help, /balance, /transfer @usuario 50, /history")
print("="*60 + "\n")

await app.initialize()
await app.start()
await app.updater.start_polling()
```
